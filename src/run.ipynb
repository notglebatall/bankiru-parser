```python
import pandas as pd

from IPython.display import display

from banki_ru.deposits import collect_deposits
from banki_ru.savings import collect_savings

from bankiros.deposits import collect_deposits as collect_bankiros_deposits
from sravni.deposits import collect_deposits as collect_sravni_deposits
```

### banki.ru

```python
rates = collect_deposits()
```

```python
print(f"Банков: {len({row['bank_id'] for row in rates})}; вкладов: {len(rates)}")
with pd.option_context("display.max_rows", None, "display.max_columns", None):
    display(pd.DataFrame(rates).head(15))
```

```python
savings_rates = collect_savings()
```

```python
print(f"Банков: {len({row['bank_id'] for row in savings_rates})}; накопительных счетов: {len(savings_rates)}")
with pd.option_context("display.max_rows", None, "display.max_columns", None):
    display(pd.DataFrame(savings_rates).head(15))
```
